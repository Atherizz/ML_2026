## Penjelasan Kegunaan Library

* **PyPREP**: Otomatisasi *preprocessing* sinyal EEG (Electroencephalography) untuk mendeteksi kanal rusak (*bad channels*), interpolasi data, dan kalkulasi *robust average reference*.
* **SciPy**: Komputasi saintifik dan matematika tingkat lanjut, terutama modul `scipy.signal` untuk pemrosesan sinyal digital (*filtering*, FFT, deteksi puncak gelombang).
* **wandb (Weights & Biases)**: Platform MLOps untuk *experiment tracking*, pencatatan log metrik performa model secara *real-time*, *hyperparameter tuning*, dan manajemen *artifacts*.
* **pyECG**: Ekstraksi dan pembacaan format data sinyal rekam jantung (ECG/EKG), seperti file PhysioBank/WFDB dan Holter menjadi array numerik.
