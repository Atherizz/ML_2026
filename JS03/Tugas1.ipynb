```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/Atherizz/ML_2026/main/JS03/wbc.csv'
df = pd.read_csv(url)

# Tampilkan 5 data teratas
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

#### Langkah 1: Pemisahan Variabel

```python
df = df.drop(columns=['id', 'Unnamed: 32'])
```

#### Langkah 2: Encoding Label Target

```python
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
```

```python
from sklearn.preprocessing import StandardScaler
```

#### Langkah 3: Standarisasi Fitur Numerik

```python
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
```

#### Langkah 4: Stratified Split Data (80:20)

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
        X_scaled, y, test_size=0.2, random_state=42, stratify=y
    )

print("Distribusi kelas data asli:\n", y.value_counts(normalize=True))
print("\nDistribusi kelas Train (80%):\n", y_train.value_counts(normalize=True))
print("\nDistribusi kelas Test (20%):\n", y_test.value_counts(normalize=True))
```
